## Diagnosis 1: The Interval That Under-Covers (18 min)

An analyst reports a coefficient with a tight interval. The specification is
correct, the coefficient is unbiased, and the interval is wrong.

To see how wrong, you need many samples from a population whose truth you know, and a count of
how often the interval contains that truth: its **coverage**. A 95% interval should contain it in
95% of samples.

**Reading the code.** `np.random.default_rng(seed)` makes a random-number generator; the same
seed gives the same draws. `r.uniform(1, 10, n)` draws n numbers spread evenly between 1 and 10,
and `r.normal(0, 0.5 * x, n)` draws n normal errors whose standard deviation is 0.5 times each
row's x, so the spread grows with x: heteroskedasticity. `sm.add_constant(x)` adds a column of
ones, the intercept, and `sm.OLS(y, X).fit()` fits the regression, outcome first.

````python
# GIVEN — run as-is
# Diagnosis 1: one sample of n rows in which the error's spread grows with x
TRUE_SLOPE = 2.0


def one_trial(seed, n=400):
    """Draw one sample and fit OLS of y on x; the true slope is TRUE_SLOPE."""
    r = np.random.default_rng(seed)
    x = r.uniform(1, 10, n)
    y = 5 + TRUE_SLOPE * x + r.normal(0, 0.5 * x, n)     # variance grows with x
    return sm.OLS(y, sm.add_constant(x)).fit()
````

**The coverage count.** The next cell draws 500 samples, each with its own seed, and builds three
95% intervals for the slope in each. `fit.get_robustcov_results(cov_type="HC1")` gives back the
same fit with HC1 standard errors, and so HC1 intervals; the estimate does not change.
`.conf_int()` gives one row per coefficient, its lower and upper limit, and row 1 is the slope.
`lo <= TRUE_SLOPE <= hi` is True when the interval contains the truth, and adding True to a count
adds 1. `hits` is a dictionary, one count per name, read and updated as `hits[name]`.

The **MC SE** column is the Monte Carlo standard error of each coverage share, √(p(1 − p) / 500):
how far the share could move by chance with 500 samples. `{p:>11.1%}` prints p as a percentage
with one decimal, right-aligned in 11 characters.

````python
# GIVEN — run as-is
# Diagnosis 1, continued: in 500 samples, how often does each 95% interval contain the truth?
TRIALS = 500
hits = {"classical": 0, "HC1": 0, "HC3": 0}      # intervals that contained the true slope
for i in range(TRIALS):
    fit = one_trial(2000 + i)
    for name in hits:
        result = fit if name == "classical" else fit.get_robustcov_results(cov_type=name)
        lo, hi = result.conf_int()[1]            # row 1: the slope's interval
        hits[name] += lo <= TRUE_SLOPE <= hi     # True counts as 1

print(f"{'cov type':>12}{'coverage':>11}{'MC SE':>9}")
print("-" * 32)
for name, count in hits.items():
    p = count / TRIALS
    mc_se = (p * (1 - p) / TRIALS) ** 0.5
    print(f"{name:>12}{p:>11.1%}{mc_se:>9.1%}")
````

**Expected output:** against a nominal 95%, coverage of 92.4% with classical standard errors,
94.2% with HC1 and 94.4% with HC3, and Monte Carlo SEs of 1.2%, 1.0% and 1.0%.

### YOUR DIAGNOSIS

1. The estimate is unbiased throughout. Explain precisely what is broken and why it is the interval rather than the point estimate.
2. Is classical's shortfall from 95% more than two Monte Carlo SEs? Is the HC1-HC3 gap? Then say which of the three intervals head to 95% as n grows, which does not, and why.

*Your diagnosis here:*
1. The interval under classic coverage is broken, because it is based off an SE that assumes homoskedasticity. The classical SE calculates it by averaging all squared residuals, then dividing it by the total spread of the regressor. Averaging only works if the variance is constant, because it works on the assumption that all data points carry the same information on the spread of the data. If they don't, like under heteroskedasticity, then you are applying an incorrect assumption to the SE calculation, which the interval is based off of. The point estimate is still unbroken because it is only concerned with how the regressor affects the outcome variable. While heteroskedasticity means that the variance is not constant, it does not systematically push the outcome variable one way or another.
2. The gap with the classical SE is 2.6 PP away from the nominal 95%, more than two times the Monte Carlo SE of 1.2*2 = 2.4 PP. The gap betwen HC1 and HC3 is 0.2 PP, less than twice the Monte Carlo SE of 2 PP. HC1 and HC3, rather than using the average of squared residuals to calculate SE, use the sum of each points squared residual weighed with its squared distance from the average x. There will still be some error in the guess of each point's noise, which is why the coverage isn't exactly 95%, but as N grows, the error in each guess is averaged out more and more, and the coverage approaches 95%. With the classical SE calculation, it wil lnot approach 95%, because it's issue is not based on noise that can be reduced with more data points. Because using the classical SE on heteroskedastic data rests on an incorrect assumption about the variance, reducing the noise of the data only gives you a more precise estimate of the wrong number.

## Diagnosis 2: The Violation Robust Errors Do Not Fix (25 min)

Many analysts switch on robust standard errors by habit, as if they fixed every problem with the
errors. This diagnosis tests that habit on errors that are shared within a firm.

The next cell builds a panel of 40 firms with 25 employees each. Everyone in a firm shares one
shock to the outcome, and x has a part shared within the firm as well, as training budgets and
wages would.

**Reading the code.** `np.repeat(np.arange(40), 25)` is 0 repeated 25 times, then 1 repeated 25
times, and so on up to 39: each row's firm. `r.normal(0, 1.2, N_FIRMS)` draws one shock per firm,
and `[firm]` copies each firm's draw to all 25 of its rows. `pd.DataFrame({...})` builds a table
from named columns.

````python
# GIVEN — run as-is
# Diagnosis 2: 40 firms, 25 employees each, error shared within firm
N_FIRMS, PER = 40, 25
TRUE_B = 1.5


def clustered_panel(seed):
    """One panel: each row's firm, x, and y = 1.5x + a firm shock + an individual error."""
    r = np.random.default_rng(seed)
    firm = np.repeat(np.arange(N_FIRMS), PER)
    firm_shock = r.normal(0, 1.2, N_FIRMS)[firm]     # shared within firm
    x = r.normal(0, 1, N_FIRMS * PER) + r.normal(0, 1, N_FIRMS)[firm]   # part own, part firm
    y = TRUE_B * x + firm_shock + r.normal(0, 0.6, N_FIRMS * PER)
    return pd.DataFrame({"firm": firm, "x": x, "y": y})


panel = clustered_panel(RANDOM_STATE)
````

**Three standard errors for one slope.** `.fit(cov_type="HC3")` fits the same regression with HC3
standard errors, as `get_robustcov_results` did in Diagnosis 1. `cov_type="cluster"` with
`cov_kwds={"groups": panel["firm"]}` clusters them by firm: it lets the errors of rows in the same
firm be correlated. `fits` is a dictionary of the three fits, and `.items()` walks through its
names and fits in pairs. `.params.iloc[1]` is the second coefficient, the slope (`.iloc` picks by
position), and `.bse` holds the standard errors.

````python
# GIVEN — run as-is
# Diagnosis 2, continued: one regression, three standard errors for its slope
base = sm.OLS(panel["y"], sm.add_constant(panel["x"]))
fits = {
    "classical": base.fit(),
    "HC3 robust": base.fit(cov_type="HC3"),
    "clustered by firm": base.fit(cov_type="cluster",
                                  cov_kwds={"groups": panel["firm"]}),
}
print(f"{'cov type':>20}{'estimate':>11}{'std err':>10}")
print("-" * 42)
for name, f in fits.items():
    print(f"{name:>20}{f.params.iloc[1]:>11.4f}{f.bse.iloc[1]:>10.4f}")
````

**Expected output:** the estimate is 1.4927 in all three rows; the standard errors are 0.0254
(classical), 0.0264 (HC3) and 0.0879 (clustered by firm).

**Which one actually covers?** The next cell repeats Diagnosis 1's count on 300 new panels, each
with its own seed. With a DataFrame as input, `.conf_int()` gives a table, so `.iloc[1]` picks the
slope's row. It takes a few seconds.

````python
# GIVEN — run as-is
# Diagnosis 2, continued: which interval actually covers?
TRIALS = 300
covered = {"classical": 0, "HC3": 0, "cluster": 0}     # intervals that contained TRUE_B
for i in range(TRIALS):
    sim_panel = clustered_panel(7000 + i)
    sim_model = sm.OLS(sim_panel["y"], sm.add_constant(sim_panel["x"]))
    for name, fit in [("classical", sim_model.fit()),
                      ("HC3", sim_model.fit(cov_type="HC3")),
                      ("cluster", sim_model.fit(cov_type="cluster",
                                                cov_kwds={"groups": sim_panel["firm"]}))]:
        lo, hi = fit.conf_int().iloc[1]
        covered[name] += lo <= TRUE_B <= hi

for name, count in covered.items():
    p = count / TRIALS
    print(f"{name:>10}: {p:6.1%}  (MC SE {(p*(1-p)/TRIALS)**0.5:.1%})   nominal 95%")
````

**Expected output:** classical 49.0% (MC SE 2.9%), HC3 50.3% (MC SE 2.9%), cluster 93.7% (MC SE
1.4%).

## Ship It: `src/diagnostics.py` (7 min)

Here is the finished module, reading only: run the two cells and compare the output with the
expected output below. `recommend_cov` gives back the covariance type to use and the reason,
decided from the design, not from a test on the residuals: HC3 when the rows are independent,
clustered errors when you pass the groups, and a warning below 40 clusters (Chapter 14).
`fit_as_recommended` fits OLS with that choice.

`%%writefile src/diagnostics.py` saves the cell as that file instead of running it. `groups=None`
makes `groups` optional, `-> dict` only notes the type given back, and `.nunique()` counts the
distinct groups.

````python
%%writefile src/diagnostics.py
"""Regression diagnostics that end in a recommendation (ECON 5200 Lab 14).

The design rule: a diagnostic that reports a statistic and stops leaves the reader to guess, so
recommend_cov gives back the covariance type to use and the reason for it.
"""
import pandas as pd
import statsmodels.api as sm

FEW_CLUSTERS = 40      # below about 40 clusters, cluster-robust tests reject too often (Ch 14)


def recommend_cov(groups=None) -> dict:
    """The covariance type the design calls for, chosen before looking at any residuals."""
    if groups is None:
        return {"cov_type": "HC3", "cov_kwds": {},
                "reason": "rows treated as independent: HC3 by default, whatever a test says"}
    n_groups = pd.Series(groups).nunique()
    reason = f"errors may be shared within each of {n_groups} groups: cluster on them"
    if n_groups < FEW_CLUSTERS:
        reason += f"; only {n_groups} clusters, so check with a wild cluster bootstrap"
    return {"cov_type": "cluster", "cov_kwds": {"groups": groups}, "reason": reason}


def fit_as_recommended(y, X, groups=None):
    """OLS of y on X plus an intercept, with the recommended covariance type, and the advice."""
    advice = recommend_cov(groups)
    fit = sm.OLS(y, sm.add_constant(X)).fit(cov_type=advice["cov_type"],
                                            cov_kwds=advice["cov_kwds"])
    return fit, advice
````

*Your prompt here:* I am an economics student writing in Python using numpy as np, pandas as pd, and statsmodels.api as sm. Write a function se_table(df, outcome, regressors, cluster) where df is a dataframe of the dataset, outcome is the name of the outcome column in the dataset, regressors is a list of regressor column names in the dataset, and cluster is the name of the clustering column. Each row must have observations for all regressors and the outcome variable, so for any row that is missing an observation in one or more of those columns, the function should drop that row using .dropna(). The function should then run a regression with sm.OLS using the outcome and regressor column names, with an intercept added with sm.add_constant(). The function should make a dataframe containing 6 different kinds of statsmodels standard errors stored in .bse of the fitted model: the standard SE labelled as classical, the heteroskedastic robust standard errors HC0, HC1, HC2, and HC3, each labelled as HC0, HC1, HC2, and HC3 respectively, and finally clustered standard error, clustering on the cluster column name, and labelled "cluster". The function should use the default small-sample corrections in statsmodels, rather than inventing any by itself. The function should return 2 things, first it should return an indication of how many rows out of the total were used in the regression after the dropping step. Second, it should return the dataframe of the different standard errors as a table. The table should have a row for each coefficient in the model, starting with the intercept and calling it "const". For each row, there should be a column that reports the SE of the coefficient for each of the 6 types of SE, and each one should be converted to a string floated to 5 decimal points to make it legible while keeping the full values saved. Include a docstring that explains what the function does and returns, and comment each line to explain what the function is doing.

````python
# Paste the se_table function from round 1 here, then run the check below.
# Paste only the function: if the AI also wrote example data or fitted a model, leave it out.

def se_table(df, outcome, regressors, cluster):
    """
    Fit a single OLS regression and compare six different standard errors for it.

    The function first drops every row that is missing the outcome or any of the
    regressors, so that all six specifications are estimated on exactly the same
    sample. It then fits y on a constant plus the regressors with sm.OLS, and
    re-fits that same model under six covariance assumptions, pulling the
    standard errors out of .bse each time:

        classical : the usual homoskedastic OLS standard errors
        HC0       : White heteroskedasticity-robust, no small-sample correction
        HC1       : HC0 scaled by n / (n - k)
        HC2       : HC0 with residuals scaled by 1 / (1 - h_ii)
        HC3       : HC0 with residuals scaled by 1 / (1 - h_ii)^2
        cluster   : clustered on the `cluster` column

    All small-sample corrections are whatever statsmodels applies by default
    (including the default finite-sample correction for the clustered fit);
    nothing is rescaled by hand here.

    Parameters
    ----------
    df : pandas.DataFrame
        The dataset.
    outcome : str
        Name of the outcome (dependent variable) column.
    regressors : list of str
        Names of the regressor columns.
    cluster : str
        Name of the column whose values define the clusters.

    Returns
    -------
    rows_used : str
        A short message reporting how many rows of the original dataframe were
        used in the regression after dropping incomplete rows.
    se_strings : pandas.DataFrame
        A table with one row per coefficient (the intercept, labelled "const",
        first) and one column per standard error type. Entries are strings
        formatted to 5 decimal places for legibility; the underlying
        full-precision numeric table is kept in se_strings.attrs["full"].
    """
    # Count the rows in the dataframe before anything is dropped, for the report.
    n_total = len(df)

    # Build the list of columns that must be non-missing for a row to be usable.
    needed = [outcome] + list(regressors)

    # Drop any row missing the outcome or one of the regressors; other columns are kept.
    clean = df.dropna(subset=needed)

    # Count the rows that survived the drop; this is the regression sample size.
    n_used = len(clean)

    # Pull the outcome column out of the cleaned data as the dependent variable.
    y = clean[outcome]

    # Pull the regressor columns and prepend a column of ones, which statsmodels names "const".
    X = sm.add_constant(clean[list(regressors)])

    # Set up the OLS model once; every fit below re-uses these same y and X.
    model = sm.OLS(y, X)

    # Fit with the default homoskedastic covariance estimator.
    fit_classical = model.fit()

    # Re-fit asking for White's HC0 robust covariance.
    fit_hc0 = model.fit(cov_type="HC0")

    # Re-fit with HC1, the degrees-of-freedom-corrected version of HC0.
    fit_hc1 = model.fit(cov_type="HC1")

    # Re-fit with HC2, which corrects residuals using the leverage values.
    fit_hc2 = model.fit(cov_type="HC2")

    # Re-fit with HC3, the more conservative leverage correction.
    fit_hc3 = model.fit(cov_type="HC3")

    # Re-fit clustering on the cluster column, taking the groups from the cleaned sample
    # so the group labels line up row-for-row with y and X.
    fit_cluster = model.fit(cov_type="cluster", cov_kwds={"groups": clean[cluster]})

    # Collect the .bse series from each fit into one table; the shared index of
    # coefficient names ("const" first, then the regressors) becomes the row index.
    se_full = pd.DataFrame(
        {
            "classical": fit_classical.bse,
            "HC0": fit_hc0.bse,
            "HC1": fit_hc1.bse,
            "HC2": fit_hc2.bse,
            "HC3": fit_hc3.bse,
            "cluster": fit_cluster.bse,
        }
    )

    # Make a display copy in which every standard error is a string with 5 decimals.
    se_strings = se_full.apply(lambda col: col.map(lambda v: f"{v:.5f}"))

    # Stash the full-precision numeric table on the result so no information is lost.
    se_strings.attrs["full"] = se_full

    # Write the sample-size report as a readable one-line string.
    rows_used = (
        f"{n_used} of {n_total} rows used in the regression "
        f"({n_total - n_used} dropped for missing values)"
    )

    # Hand back the sample-size message and the formatted standard error table.
    return rows_used, se_strings
````

````python
# Paste the se_table function from round 2 here, then run the check below.
# Paste only the function: if the AI also wrote example data or fitted a model, leave it out.




def se_table(df, outcome, regressors, cluster):
    """
    Compare standard errors from a single OLS regression across six covariance estimators.

    The function drops every row of `df` that is missing a value for the outcome or for any
    regressor, fits one OLS model of `outcome` on `regressors` plus an intercept, and then
    re-computes the covariance matrix of that same fitted model six different ways:
    classical (homoskedastic), the four heteroskedasticity-robust estimators HC0, HC1, HC2
    and HC3, and standard errors clustered on `cluster`. All six use the small-sample
    corrections that statsmodels applies by default; nothing is adjusted by hand. The
    coefficient estimates are identical across all six columns — only the standard errors
    differ.

    Parameters
    ----------
    df : pandas.DataFrame
        The dataset.
    outcome : str
        Name of the dependent variable column in `df`.
    regressors : list of str
        Names of the independent variable columns in `df`.
    cluster : str
        Name of the column in `df` whose values define the clusters.

    Returns
    -------
    pandas.DataFrame
        A table with one row per coefficient, indexed by coefficient name with the
        intercept first and labelled "const", and six columns named "classical", "HC0",
        "HC1", "HC2", "HC3" and "cluster". Every entry is the full-precision float stored
        in the fitted model's `.bse` for that estimator.

    Side effects
    ------------
    Prints how many rows of the original dataset were used in the regression after
    dropping rows with missing values.
    """
    regressors = list(regressors)  # copy to a list so a tuple or Index argument also works
    needed = [outcome] + regressors  # the columns that must be non-missing in every used row
    n_total = len(df)  # number of rows in the dataset before any dropping
    data = df.dropna(subset=needed)  # drop any row missing the outcome or any regressor
    n_used = len(data)  # number of rows that survive and therefore enter the regression

    # Report the sample size actually used relative to the full dataset.
    print(f"Regression used {n_used} of {n_total} rows ({n_total - n_used} dropped for missing values).")

    y = data[outcome]  # dependent variable as a Series
    X = sm.add_constant(data[regressors])  # regressor matrix with a leading column of ones named "const"

    model = sm.OLS(y, X)  # set up the OLS problem once; each .fit() below reuses this same specification

    fits = {  # map each label to a fit of the same model under a different covariance estimator
        "classical": model.fit(),  # default homoskedastic (non-robust) standard errors
        "HC0": model.fit(cov_type="HC0"),  # White's original heteroskedasticity-robust estimator
        "HC1": model.fit(cov_type="HC1"),  # HC0 with the n / (n - k) degrees-of-freedom correction
        "HC2": model.fit(cov_type="HC2"),  # residuals rescaled by 1 - h_ii (leverage adjustment)
        "HC3": model.fit(cov_type="HC3"),  # residuals rescaled by (1 - h_ii)^2, the most conservative
        "cluster": model.fit(  # cluster-robust standard errors
            cov_type="cluster",  # request the one-way clustered covariance estimator
            cov_kwds={"groups": data[cluster]},  # cluster identifiers, aligned row-for-row with y and X
        ),  # small-sample corrections left at the statsmodels defaults
    }

    table = pd.DataFrame(  # assemble one column of standard errors per estimator
        {label: fit.bse for label, fit in fits.items()}  # .bse is a Series indexed by coefficient name
    )

    table = table.loc[fits["classical"].params.index]  # force the model's own coefficient order, "const" first

    return table  # the completed standard-error comparison table
````
